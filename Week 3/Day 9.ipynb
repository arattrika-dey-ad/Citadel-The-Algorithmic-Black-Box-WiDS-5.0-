```python
"""
PHASE III: Hyperparameter Tuning for Trading RL Agent - Jupyter Notebook Version
"""

# First, let's install any missing packages
import sys
import subprocess
import importlib

def install_package(package):
    """Install a package if not available"""
    try:
        importlib.import_module(package)
        print(f"✅ {package} is already installed")
        return True
    except ImportError:
        print(f"📦 Installing {package}...")
        try:
            subprocess.check_call([sys.executable, "-m", "pip", "install", package])
            print(f"✅ Successfully installed {package}")
            return True
        except:
            print(f"❌ Failed to install {package}")
            return False

# List of required packages
required_packages = [
    'gymnasium',
    'stable-baselines3',
    'optuna',
    'plotly',
    'torch'
]

print("🔍 Checking and installing required packages...")
print("-" * 50)

# Check and install packages
for package in required_packages:
    install_package(package)

print("-" * 50)
print("✅ All packages checked/installed\n")

# Now import the libraries
import os
import json
import numpy as np
import pandas as pd
from typing import Dict, Any, List, Tuple, Optional, Callable
import warnings
warnings.filterwarnings('ignore')

# Core ML/DL Libraries
import torch
import torch.nn as nn

# Use Gymnasium instead of Gym
import gymnasium as gym
from gymnasium.wrappers import RecordEpisodeStatistics
from stable_baselines3 import PPO
from stable_baselines3.common.evaluation import evaluate_policy
from stable_baselines3.common.callbacks import EvalCallback, StopTrainingOnRewardThreshold
from stable_baselines3.common.vec_env import DummyVecEnv, VecMonitor

# Optimization Framework
import optuna
from optuna.pruners import MedianPruner
from optuna.samplers import TPESampler
from optuna.trial import TrialState
from optuna.visualization import plot_optimization_history, plot_param_importances, plot_parallel_coordinate

# Visualization
import plotly.graph_objects as go
import plotly.express as px
from plotly.subplots import make_subplots
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display, HTML

# Set random seeds for reproducibility
SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed(SEED)

# ============================================================================
# 1. MOCK TRADING ENVIRONMENT (Replace with your actual environment)
# ============================================================================

class MockTradingEnv(gym.Env):
    """
    Simple trading environment for demonstration.
    Replace with your actual TradingEnvironment.
    """
    metadata = {'render_modes': ['human']}
    
    def __init__(self, render_mode=None):
        super().__init__()
        
        # Action: position size from -1 (full short) to 1 (full long)
        self.action_space = gym.spaces.Box(low=-1, high=1, shape=(1,), dtype=np.float32)
        
        # Observation: price, indicators, position, balance
        self.observation_space = gym.spaces.Box(
            low=-np.inf, high=np.inf, shape=(10,), dtype=np.float32
        )
        
        # Initialize
        self.current_step = 0
        self.max_steps = 100
        self.balance = 10000
        self.position = 0
        self.portfolio_value = self.balance
        self.render_mode = render_mode
        
        # Generate mock data
        self.generate_data()
        
    def generate_data(self):
        """Generate mock price data"""
        np.random.seed(SEED)
        self.prices = 100 + np.cumsum(np.random.randn(1000) * 0.5)
        
    def _get_observation(self):
        """Get current observation"""
        price = self.prices[self.current_step]
        # Simple indicators
        sma = np.mean(self.prices[max(0, self.current_step-20):self.current_step+1])
        rsi = 50 + np.random.randn() * 10  # Mock RSI
        
        return np.array([
            price, sma, rsi, 
            self.position, self.balance/10000, self.portfolio_value/10000,
            self.current_step/self.max_steps,
            np.random.randn(), np.random.randn(), np.random.randn()
        ], dtype=np.float32)
    
    def reset(self, seed=None, options=None):
        """Reset environment"""
        super().reset(seed=seed)
        
        self.current_step = 0
        self.balance = 10000
        self.position = 0
        self.portfolio_value = self.balance
        
        return self._get_observation(), {}
    
    def step(self, action):
        """Execute action"""
        action = float(np.clip(action[0], -1, 1))
        
        # Current and next price
        current_price = self.prices[self.current_step]
        next_idx = min(self.current_step + 1, len(self.prices) - 1)
        next_price = self.prices[next_idx]
        
        # Price change
        price_change = (next_price - current_price) / current_price
        
        # Update position
        old_position = self.position
        self.position = action
        
        # Transaction cost
        transaction_cost = abs(self.position - old_position) * current_price * 0.001
        
        # P&L from old position
        pnl = old_position * price_change * current_price
        
        # Update balance
        self.balance += pnl - transaction_cost
        
        # Update portfolio value
        self.portfolio_value = self.balance + self.position * current_price
        
        # Reward = P&L - transaction cost - risk penalty
        reward = pnl - transaction_cost - 0.01 * abs(self.position)
        
        # Update step
        self.current_step += 1
        
        # Check if done
        done = self.current_step >= self.max_steps or self.current_step >= len(self.prices) - 1
        
        # Next observation
        next_obs = self._get_observation()
        
        # Info
        info = {
            'portfolio_value': self.portfolio_value,
            'balance': self.balance,
            'position': self.position,
            'price': current_price
        }
        
        truncated = False
        
        return next_obs, reward, done, truncated, info
    
    def render(self):
        """Render environment state"""
        if self.render_mode == 'human':
            print(f"Step: {self.current_step}, "
                  f"Portfolio: ${self.portfolio_value:.2f}, "
                  f"Position: {self.position:.2f}, "
                  f"Price: ${self.prices[self.current_step]:.2f}")
    
    def close(self):
        pass

# ============================================================================
# 2. ENVIRONMENT CREATION FUNCTIONS
# ============================================================================

def make_env(env_class=MockTradingEnv, seed=0, rank=0):
    """Create a single environment instance"""
    def _init():
        env = env_class()
        env = RecordEpisodeStatistics(env)
        env.reset(seed=seed + rank)
        env.action_space.seed(seed + rank)
        return env
    return _init

def make_vec_env(n_envs=2, env_class=MockTradingEnv, seed=SEED):
    """Create vectorized environment"""
    env_fns = [make_env(env_class, seed, i) for i in range(n_envs)]
    env = DummyVecEnv(env_fns)
    env = VecMonitor(env)
    return env

# ============================================================================
# 3. OPTIMIZATION CONFIGURATION
# ============================================================================

class OptimizationConfig:
    """Configuration for hyperparameter optimization"""
    def __init__(self):
        # Study settings
        self.n_trials = 20  # Reduced for faster testing in notebook
        self.n_startup_trials = 3
        self.total_timesteps = 50000  # Per trial
        self.n_envs = 2
        self.seed = SEED
        
        # Evaluation
        self.eval_freq = 5000
        self.n_eval_episodes = 3
        
        # Hyperparameter bounds (based on RL Baselines3 Zoo)
        self.hp_bounds = {
            'learning_rate': ('log', 1e-5, 1e-3),
            'gamma': ('float', 0.9, 0.9999),
            'ent_coef': ('log', 1e-8, 0.1),
            'clip_range': ('float', 0.1, 0.3),
            'gae_lambda': ('float', 0.8, 0.99),
            'max_grad_norm': ('float', 0.3, 5.0),
            'n_steps': ('int', 64, 2048),
            'batch_size': ('categorical', [64, 128, 256, 512]),
            'n_epochs': ('int', 1, 20)
        }
        
        # Output paths
        self.log_dir = "./optuna_logs/"
        self.best_model_dir = "./best_models/"
        
    def create_directories(self):
        """Create output directories"""
        os.makedirs(self.log_dir, exist_ok=True)
        os.makedirs(self.best_model_dir, exist_ok=True)
        os.makedirs(f"{self.best_model_dir}/best_overall", exist_ok=True)

# ============================================================================
# 4. OBJECTIVE FUNCTION
# ============================================================================

class OptimizationObjective:
    """Objective function for Optuna optimization"""
    
    def __init__(self, config):
        self.config = config
        self.best_score = -np.inf
        self.results = []
        
    def __call__(self, trial):
        """Run a single optimization trial"""
        try:
            # Sample hyperparameters
            params = self._sample_params(trial)
            
            # Create environments
            train_env = make_vec_env(self.config.n_envs, MockTradingEnv, self.config.seed)
            eval_env = make_vec_env(1, MockTradingEnv, self.config.seed + 1000)
            
            # Create model
            model = PPO(
                policy='MlpPolicy',
                env=train_env,
                learning_rate=params['learning_rate'],
                gamma=params['gamma'],
                ent_coef=params['ent_coef'],
                clip_range=params['clip_range'],
                gae_lambda=params['gae_lambda'],
                max_grad_norm=params['max_grad_norm'],
                n_steps=params['n_steps'],
                batch_size=params['batch_size'],
                n_epochs=params['n_epochs'],
                verbose=0,
                device='auto',
                tensorboard_log=f"./tensorboard/trial_{trial.number}"
            )
            
            # Evaluation callback
            eval_callback = EvalCallback(
                eval_env,
                best_model_save_path=f"{self.config.best_model_dir}/trial_{trial.number}",
                eval_freq=max(self.config.eval_freq // self.config.n_envs, 1),
                n_eval_episodes=self.config.n_eval_episodes,
                deterministic=True,
                verbose=0
            )
            
            # Train model
            model.learn(
                total_timesteps=self.config.total_timesteps,
                callback=eval_callback,
                progress_bar=False
            )
            
            # Final evaluation
            mean_reward, std_reward = evaluate_policy(
                model, eval_env, n_eval_episodes=5, deterministic=True
            )
            
            # Calculate score (mean reward penalized by variance)
            score = mean_reward / (1 + abs(std_reward))
            
            # Store results
            trial_result = {
                'trial': trial.number,
                'params': params,
                'score': score,
                'mean_reward': mean_reward,
                'std_reward': std_reward
            }
            self.results.append(trial_result)
            
            # Save if best
            if score > self.best_score:
                self.best_score = score
                model.save(f"{self.config.best_model_dir}/best_overall/model.zip")
                
                # Save metadata
                metadata = {
                    'trial': trial.number,
                    'score': score,
                    'params': params,
                    'mean_reward': mean_reward,
                    'std_reward': std_reward,
                    'timestamp': pd.Timestamp.now().isoformat()
                }
                with open(f"{self.config.best_model_dir}/best_overall/metadata.json", 'w') as f:
                    json.dump(metadata, f, indent=2)
            
            # Cleanup
            train_env.close()
            eval_env.close()
            del model
            
            return score
            
        except Exception as e:
            print(f"❌ Trial {trial.number} failed: {str(e)[:100]}...")
            return -1000  # Very low score for failed trials
    
    def _sample_params(self, trial):
        """Sample hyperparameters from distributions"""
        params = {}
        
        for param, (param_type, *bounds) in self.config.hp_bounds.items():
            if param_type == 'log':
                params[param] = trial.suggest_float(param, bounds[0], bounds[1], log=True)
            elif param_type == 'float':
                params[param] = trial.suggest_float(param, bounds[0], bounds[1])
            elif param_type == 'int':
                params[param] = trial.suggest_int(param, bounds[0], bounds[1])
            elif param_type == 'categorical':
                params[param] = trial.suggest_categorical(param, bounds[0])
        
        return params

# ============================================================================
# 5. VISUALIZATION FUNCTIONS
# ============================================================================

def plot_study_results(study):
    """Create interactive visualizations of study results"""
    
    # Create a 2x2 grid of plots
    fig = make_subplots(
        rows=2, cols=2,
        subplot_titles=(
            'Optimization History',
            'Parameter Importances',
            'Learning Rate vs Score',
            'Gamma vs Score'
        ),
        specs=[[{"type": "scatter"}, {"type": "bar"}],
               [{"type": "scatter"}, {"type": "scatter"}]]
    )
    
    # 1. Optimization History
    trials_df = study.trials_dataframe()
    if not trials_df.empty:
        completed = trials_df[trials_df['state'] == 'COMPLETE']
        if not completed.empty:
            fig.add_trace(
                go.Scatter(
                    x=completed['number'],
                    y=completed['value'],
                    mode='lines+markers',
                    name='Objective',
                    line=dict(color='blue', width=2),
                    marker=dict(size=6)
                ),
                row=1, col=1
            )
    
    # 2. Parameter Importances
    try:
        importances = optuna.importance.get_param_importances(study)
        fig.add_trace(
            go.Bar(
                x=list(importances.keys()),
                y=list(importances.values()),
                name='Importance',
                marker_color='lightgreen'
            ),
            row=1, col=2
        )
    except:
        fig.add_annotation(
            text="Need more trials for importance analysis",
            xref="x2", yref="y2",
            x=0.5, y=0.5,
            showarrow=False,
            row=1, col=2
        )
    
    # 3. Learning Rate vs Score
    if not trials_df.empty and 'params_learning_rate' in trials_df.columns:
        fig.add_trace(
            go.Scatter(
                x=trials_df['params_learning_rate'],
                y=trials_df['value'],
                mode='markers',
                name='LR vs Score',
                marker=dict(
                    size=8,
                    color=trials_df['value'],
                    colorscale='Viridis',
                    showscale=True
                )
            ),
            row=2, col=1
        )
    
    # 4. Gamma vs Score
    if not trials_df.empty and 'params_gamma' in trials_df.columns:
        fig.add_trace(
            go.Scatter(
                x=trials_df['params_gamma'],
                y=trials_df['value'],
                mode='markers',
                name='Gamma vs Score',
                marker=dict(
                    size=8,
                    color=trials_df['params_learning_rate'],
                    colorscale='Plasma',
                    showscale=True,
                    colorbar=dict(title='Learning Rate')
                )
            ),
            row=2, col=2
        )
    
    # Update layout
    fig.update_layout(
        height=800,
        title_text="Hyperparameter Optimization Results",
        showlegend=True,
        template="plotly_white"
    )
    
    # Update axes labels
    fig.update_xaxes(title_text="Trial", row=1, col=1)
    fig.update_yaxes(title_text="Objective Value", row=1, col=1)
    fig.update_xaxes(title_text="Parameter", row=1, col=2)
    fig.update_yaxes(title_text="Importance", row=1, col=2)
    fig.update_xaxes(title_text="Learning Rate", row=2, col=1)
    fig.update_yaxes(title_text="Score", row=2, col=1)
    fig.update_xaxes(title_text="Gamma", row=2, col=2)
    fig.update_yaxes(title_text="Score", row=2, col=2)
    
    return fig

def print_study_summary(study):
    """Print a comprehensive summary of the study"""
    print("=" * 80)
    print("📊 OPTIMIZATION STUDY SUMMARY")
    print("=" * 80)
    
    # Basic statistics
    n_trials = len(study.trials)
    n_completed = len([t for t in study.trials if t.state == TrialState.COMPLETE])
    n_pruned = len([t for t in study.trials if t.state == TrialState.PRUNED])
    
    print(f"\n📈 TRIAL STATISTICS:")
    print(f"   Total Trials: {n_trials}")
    print(f"   Completed: {n_completed} ({n_completed/n_trials:.1%})")
    print(f"   Pruned: {n_pruned} ({n_pruned/n_trials:.1%})")
    
    if study.best_trial:
        print(f"\n🎯 BEST TRIAL:")
        print(f"   Trial Number: {study.best_trial.number}")
        print(f"   Objective Value: {study.best_value:.4f}")
        
        print(f"\n⚙️  BEST HYPERPARAMETERS:")
        for param, value in study.best_params.items():
            print(f"   {param:20s}: {value}")
    
    # Parameter ranges analysis
    if n_completed > 3:
        trials_df = study.trials_dataframe()
        completed = trials_df[trials_df['state'] == 'COMPLETE']
        
        print(f"\n🔍 PARAMETER ANALYSIS:")
        
        # Check for boundary conditions
        bounds = {
            'learning_rate': (1e-5, 1e-3),
            'gamma': (0.9, 0.9999),
            'ent_coef': (1e-8, 0.1)
        }
        
        for param, (low, high) in bounds.items():
            if f'params_{param}' in completed.columns:
                median_val = completed[f'params_{param}'].median()
                if abs(median_val - low) / (high - low) < 0.1:
                    print(f"   ⚠️  {param}: Median near lower bound ({median_val:.2e})")
                elif abs(median_val - high) / (high - low) < 0.1:
                    print(f"   ⚠️  {param}: Median near upper bound ({median_val:.2e})")

# ============================================================================
# 6. MAIN OPTIMIZATION FUNCTION
# ============================================================================

def run_hyperparameter_optimization(n_trials=20, total_timesteps=50000):
    """
    Run hyperparameter optimization in Jupyter notebook
    
    Args:
        n_trials: Number of optimization trials
        total_timesteps: Training steps per trial
    """
    print("=" * 80)
    print("🚀 STARTING HYPERPARAMETER OPTIMIZATION")
    print("=" * 80)
    print(f"📊 Configuration:")
    print(f"   Trials: {n_trials}")
    print(f"   Timesteps per trial: {total_timesteps:,}")
    print(f"   Environment: MockTradingEnv")
    print("=" * 80)
    
    # Create configuration
    config = OptimizationConfig()
    config.n_trials = n_trials
    config.total_timesteps = total_timesteps
    config.create_directories()
    
    # Create study
    sampler = TPESampler(seed=SEED)
    pruner = MedianPruner()
    
    study = optuna.create_study(
        study_name="trading_ppo_opt",
        sampler=sampler,
        pruner=pruner,
        direction="maximize",
        storage=None  # In-memory storage for notebook
    )
    
    # Create objective
    objective = OptimizationObjective(config)
    
    # Run optimization
    print("\n🎯 Starting optimization trials...")
    print("   (This may take a while depending on n_trials and timesteps)")
    print("-" * 50)
    
    study.optimize(
        objective,
        n_trials=n_trials,
        show_progress_bar=True,
        n_jobs=1  # Set to 1 for stability in notebook
    )
    
    print("-" * 50)
    print("✅ Optimization completed!")
    
    # Save results
    trials_df = study.trials_dataframe()
    if not trials_df.empty:
        trials_df.to_csv(f"{config.log_dir}/trials.csv", index=False)
        print(f"💾 Results saved to {config.log_dir}/trials.csv")
    
    # Generate visualizations
    print("\n📊 Generating visualizations...")
    fig = plot_study_results(study)
    fig.write_html(f"{config.log_dir}/optimization_results.html")
    
    # Display summary
    print_study_summary(study)
    
    # Display the plot
    display(fig)
    
    return study, objective

# ============================================================================
# 7. TEST BEST MODEL FUNCTION
# ============================================================================

def test_best_model():
    """Load and test the best model from optimization"""
    print("=" * 80)
    print("🧪 TESTING BEST MODEL")
    print("=" * 80)
    
    model_path = "./best_models/best_overall/model.zip"
    
    if not os.path.exists(model_path):
        print("❌ No best model found. Run optimization first.")
        return None
    
    # Load model
    model = PPO.load(model_path)
    print("✅ Model loaded successfully")
    
    # Load metadata
    with open("./best_models/best_overall/metadata.json", 'r') as f:
        metadata = json.load(f)
    
    print(f"\n📋 Model Info:")
    print(f"   Trial: {metadata['trial']}")
    print(f"   Score: {metadata['score']:.4f}")
    print(f"   Mean Reward: {metadata['mean_reward']:.2f}")
    
    # Create test environment
    test_env = make_vec_env(1, MockTradingEnv, SEED + 999)
    
    # Run a few episodes
    print("\n🎮 Running test episodes...")
    all_rewards = []
    
    for episode in range(3):
        obs = test_env.reset()
        done = [False]
        total_reward = 0
        step = 0
        
        while not done[0] and step < 100:
            action, _ = model.predict(obs, deterministic=True)
            obs, reward, done, info = test_env.step(action)
            total_reward += reward[0]
            step += 1
        
        all_rewards.append(total_reward)
        print(f"   Episode {episode + 1}: Reward = {total_reward:.2f}")
    
    print(f"\n📊 Test Statistics:")
    print(f"   Mean Reward: {np.mean(all_rewards):.2f}")
    print(f"   Std Reward: {np.std(all_rewards):.2f}")
    
    # Plot portfolio value over time
    print("\n📈 Running detailed episode...")
    obs = test_env.reset()
    done = [False]
    portfolio_values = []
    
    while not done[0] and len(portfolio_values) < 100:
        action, _ = model.predict(obs, deterministic=True)
        obs, reward, done, info = test_env.step(action)
        portfolio_values.append(info[0].get('portfolio_value', 0))
    
    # Create portfolio plot
    fig = go.Figure()
    fig.add_trace(go.Scatter(
        y=portfolio_values,
        mode='lines',
        name='Portfolio Value',
        line=dict(color='green', width=2)
    ))
    
    fig.update_layout(
        title="Portfolio Value During Test Episode",
        xaxis_title="Step",
        yaxis_title="Portfolio Value ($)",
        template="plotly_white"
    )
    
    display(fig)
    
    test_env.close()
    return model

# ============================================================================
# 8. HYPERPARAMETER ANALYSIS FUNCTION
# ============================================================================

def analyze_hyperparameters(study):
    """Analyze hyperparameter effects and provide insights"""
    print("=" * 80)
    print("🔬 HYPERPARAMETER ANALYSIS")
    print("=" * 80)
    
    trials_df = study.trials_dataframe()
    if trials_df.empty or len(trials_df) < 5:
        print("❌ Need at least 5 completed trials for analysis")
        return
    
    completed = trials_df[trials_df['state'] == 'COMPLETE']
    
    if len(completed) < 3:
        print("❌ Not enough completed trials for analysis")
        return
    
    print("\n📊 CORRELATION ANALYSIS:")
    print("-" * 40)
    
    # Calculate correlations with objective value
    param_cols = [col for col in completed.columns if col.startswith('params_')]
    
    correlations = {}
    for col in param_cols:
        corr = completed['value'].corr(completed[col])
        if not pd.isna(corr):
            param_name = col.replace('params_', '')
            correlations[param_name] = abs(corr)
    
    # Sort by absolute correlation
    sorted_corrs = sorted(correlations.items(), key=lambda x: x[1], reverse=True)
    
    for param, corr in sorted_corrs:
        strength = "STRONG" if corr > 0.5 else "MODERATE" if corr > 0.3 else "WEAK"
        print(f"   {param:20s}: {corr:.3f} ({strength})")
    
    print("\n💡 INSIGHTS & RECOMMENDATIONS:")
    print("-" * 40)
    
    # Analyze specific parameters
    insights = []
    
    # Learning rate analysis
    if 'learning_rate' in correlations:
        lr_median = completed['params_learning_rate'].median()
        if lr_median < 1e-4:
            insights.append("• Learning Rate: Very low optimal LR suggests high gradient variance")
        elif lr_median > 1e-3:
            insights.append("• Learning Rate: High optimal LR suggests stable learning dynamics")
    
    # Gamma analysis
    if 'gamma' in correlations:
        gamma_median = completed['params_gamma'].median()
        if gamma_median > 0.99:
            insights.append("• Gamma: High discount (γ≈1) → Agent benefits from long-term planning")
        elif gamma_median < 0.95:
            insights.append("• Gamma: Low discount (γ<0.95) → Agent focuses on short-term rewards")
    
    # Entropy coefficient analysis
    if 'ent_coef' in correlations:
        ent_median = completed['params_ent_coef'].median()
        if ent_median < 1e-4:
            insights.append("• Entropy: Very low → Agent quickly exploits learned policies")
        elif ent_median > 0.01:
            insights.append("• Entropy: High → Environment requires continued exploration")
    
    # N-steps analysis
    if 'n_steps' in correlations:
        nsteps_median = completed['params_n_steps'].median()
        if nsteps_median > 1000:
            insights.append("• N-steps: Large rollout (>" + str(nsteps_median) + ") → High correlation between actions")
    
    if insights:
        for insight in insights:
            print(f"   {insight}")
    else:
        print("   • No strong patterns detected. Consider expanding hyperparameter ranges.")
    
    print("\n🎯 NEXT STEPS:")
    print("-" * 40)
    
    # Check if we should expand search
    bounds_hit = []
    
    # Learning rate bounds
    if 'params_learning_rate' in completed.columns:
        if completed['params_learning_rate'].max() > 0.9e-3:
            bounds_hit.append("learning_rate (upper)")
        if completed['params_learning_rate'].min() < 1.1e-5:
            bounds_hit.append("learning_rate (lower)")
    
    if bounds_hit:
        print(f"   ⚠️  Consider expanding bounds for: {', '.join(bounds_hit)}")
    else:
        print("   ✅ Hyperparameter ranges seem appropriate")
    
    print("\n   🔄 For better optimization:")
    print("      • Increase n_trials to 50+")
    print("      • Increase total_timesteps per trial")
    print("      • Replace MockTradingEnv with your actual environment")
    print("      • Add more diverse hyperparameters (network architecture, etc.)")

# ============================================================================
# 9. NOTEBOOK INTERFACE FUNCTIONS
# ============================================================================

def run_quick_test():
    """Run a quick optimization test (3 trials, 10k steps)"""
    print("Quick test mode: 3 trials, 10,000 steps each")
    print("This will take 1-2 minutes...")
    
    study, _ = run_hyperparameter_optimization(
        n_trials=3,
        total_timesteps=10000
    )
    
    return study

def run_full_optimization():
    """Run full optimization (20 trials, 50k steps)"""
    print("Full optimization: 20 trials, 50,000 steps each")
    print("This will take 10-20 minutes...")
    
    study, objective = run_hyperparameter_optimization(
        n_trials=20,
        total_timesteps=50000
    )
    
    # Run analysis
    analyze_hyperparameters(study)
    
    return study, objective

# ============================================================================
# 10. EXAMPLE USAGE IN NOTEBOOK
# ============================================================================

# Uncomment and run these cells in your notebook:

"""
# Cell 1: Quick test (recommended first)
print("Running quick test...")
study = run_quick_test()
"""

"""
# Cell 2: Full optimization
print("Running full optimization...")
study, objective = run_full_optimization()
"""

"""
# Cell 3: Test the best model
print("Testing best model...")
model = test_best_model()
"""

"""
# Cell 4: Manual analysis
if 'study' in locals():
    analyze_hyperparameters(study)
"""

# ============================================================================
# 11. REPLACE WITH YOUR ENVIRONMENT
# ============================================================================

def use_your_environment():
    """
    Instructions to replace MockTradingEnv with your actual environment
    
    Steps:
    1. Import your TradingEnvironment class
    2. Replace MockTradingEnv with YourTradingEnv in the functions
    3. Adjust observation/action space if needed
    """
    print("=" * 80)
    print("🔄 REPLACING WITH YOUR TRADING ENVIRONMENT")
    print("=" * 80)
    
    print("\n📝 To use your actual TradingEnvironment:")
    print("-" * 50)
    
    print("1. Import your environment class:")
    print("   from your_trading_module import TradingEnvironment")
    print()
    
    print("2. Replace in make_env function:")
    print("   Change: env = MockTradingEnv()")
    print("   To:     env = TradingEnvironment(your_kwargs)")
    print()
    
    print("3. Update hyperparameter bounds if needed:")
    print("   - Adjust learning_rate based on reward scale")
    print("   - Adjust gamma based on time horizon")
    print("   - Adjust ent_coef based on exploration needs")
    print()
    
    print("4. Run optimization:")
    print("   study, objective = run_full_optimization()")
    print()
    
    print("5. Analyze results:")
    print("   analyze_hyperparameters(study)")
    print("   test_best_model()")

# ============================================================================
# DEMONSTRATION CELLS (Run these in order)
# ============================================================================

print("✅ Hyperparameter Optimization Module Loaded Successfully!")
print("\n" + "=" * 80)
print("📚 AVAILABLE FUNCTIONS:")
print("=" * 80)
print("\n1. run_quick_test() - Run 3 quick trials (1-2 minutes)")
print("2. run_full_optimization() - Run 20 trials (10-20 minutes)")
print("3. test_best_model() - Test the best saved model")
print("4. analyze_hyperparameters(study) - Analyze study results")
print("5. use_your_environment() - Instructions for using your own environment")
print("\n" + "=" * 80)
print("🚀 RECOMMENDED FIRST STEP:")
print("   study = run_quick_test()")
print("=" * 80)

# Uncomment to run a quick test automatically
# print("\nRunning quick test...")
# study = run_quick_test()
```
